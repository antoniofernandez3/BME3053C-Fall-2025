# What is GitHub Copilot?

GitHub Copilot is an AI pair programmer that helps you write code faster and with less effort. It works by:

- Suggesting whole lines or blocks of code as you type
- Converting comments into working code
- Helping you understand and work with unfamiliar libraries
- Offering multiple suggestions for you to choose from
- Learning from your coding style and preferences

# Switching AI Models in Copilot

GitHub Copilot offers different AI models to help with various coding tasks. Here's how to switch between them:

## Available Models

- **Default (GPT-3.5 Turbo)**: The standard model for general coding assistance
- **GPT-4**: More powerful model with enhanced code understanding and generation capabilities
- **Copilot Chat**: Conversational interface for complex queries and explanations

## How to Switch Models

1. **In VS Code**:
    - Click on the Copilot icon in the status bar
    - Select "GitHub Copilot: Switch Model"
    - Choose your preferred model from the dropdown menu

2. **Using Command Palette**:
    - Press `Ctrl+Shift+P` (Windows/Linux) or `Cmd+Shift+P` (Mac)
    - Type "GitHub Copilot: Switch Model"
    - Select the desired model

3. **In Copilot Chat**:
    - Type `/model` followed by the model name
    - Example: `/model gpt-4`

## When to Use Different Models

- **Default Model**: Everyday coding tasks, quick suggestions
- **GPT-4**: Complex algorithms, thorough code explanations, challenging problems
- **Specialized Models**: Domain-specific coding (when available)

Remember that more powerful models might be slower to respond or have usage limitations based on your subscription plan.

### ✏️ Exercise: Write a class for managing patient records
   - Include methods for adding, updating, and retrieving records
   - Implement data validation
   - Add methods for basic statistical analysis

````python
from dataclasses import dataclass, field, asdict
from datetime import date, datetime
from typing import Dict, List, Optional, Any
from collections import Counter
from statistics import mean, median
import copy

@dataclass
class Patient:
    patient_id: str
    name: str
    dob: date
    gender: Optional[str] = None  # 'M', 'F', 'O', or None
    weight_kg: Optional[float] = None
    conditions: List[str] = field(default_factory=list)
    extra: Dict[str, Any] = field(default_factory=dict)

    def age(self, on_date: Optional[date] = None) -> int:
        """Compute age in years at on_date (defaults to today)."""
        if on_date is None:
            on_date = date.today()
        if self.dob > on_date:
            raise ValueError("DOB cannot be in the future for age calculation.")
        years = on_date.year - self.dob.year
        if (on_date.month, on_date.day) < (self.dob.month, self.dob.day):
            years -= 1
        return years


class PatientRecordManager:
    VALID_GENDERS = {"M", "F", "O", None}

    def __init__(self):
        self._records: Dict[str, Patient] = {}

    # -------------------------
    # Validation helpers
    # -------------------------
    def _validate_patient_id(self, patient_id: str):
        if not isinstance(patient_id, str) or not patient_id.strip():
            raise ValueError("patient_id must be a non-empty string.")

    def _validate_name(self, name: str):
        if not isinstance(name, str) or not name.strip():
            raise ValueError("name must be a non-empty string.")

    def _validate_dob(self, dob: Any):
        if isinstance(dob, str):
            try:
                dob = datetime.fromisoformat(dob).date()
            except Exception:
                raise ValueError("dob string must be ISO format YYYY-MM-DD")
        if not isinstance(dob, date):
            raise ValueError("dob must be a datetime.date instance or ISO date string")
        if dob > date.today():
            raise ValueError("dob cannot be in the future")
        return dob

    def _validate_gender(self, gender: Optional[str]):
        if gender not in self.VALID_GENDERS:
            raise ValueError(f"gender must be one of {self.VALID_GENDERS}")

    def _validate_weight(self, weight: Optional[float]):
        if weight is None:
            return
        if not isinstance(weight, (int, float)) or weight <= 0:
            raise ValueError("weight_kg must be a positive number")

    def _validate_conditions(self, conditions: Any):
        if conditions is None:
            return []
        if not isinstance(conditions, (list, tuple)):
            raise ValueError("conditions must be a list of strings")
        for c in conditions:
            if not isinstance(c, str):
                raise ValueError("each condition must be a string")
        return list(conditions)

    # -------------------------
    # CRUD methods
    # -------------------------
    def add_patient(
        self,
        patient_id: str,
        name: str,
        dob: Any,
        gender: Optional[str] = None,
        weight_kg: Optional[float] = None,
        conditions: Optional[List[str]] = None,
        extra: Optional[Dict[str, Any]] = None,
    ) -> None:
        """Add a new patient. Raises ValueError on invalid input or duplicate id."""
        self._validate_patient_id(patient_id)
        if patient_id in self._records:
            raise ValueError(f"Patient with id '{patient_id}' already exists.")
        self._validate_name(name)
        dob = self._validate_dob(dob)
        self._validate_gender(gender)
        self._validate_weight(weight_kg)
        conditions = self._validate_conditions(conditions) or []
        extra = extra or {}

        patient = Patient(
            patient_id=patient_id.strip(),
            name=name.strip(),
            dob=dob,
            gender=gender,
            weight_kg=float(weight_kg) if weight_kg is not None else None,
            conditions=conditions,
            extra=extra,
        )
        self._records[patient.patient_id] = patient

    def update_patient(self, patient_id: str, **updates) -> None:
        """Partially update patient fields. Use same validation as add."""
        self._validate_patient_id(patient_id)
        if patient_id not in self._records:
            raise KeyError(f"Patient with id '{patient_id}' not found.")
        patient = self._records[patient_id]

        if "name" in updates:
            self._validate_name(updates["name"])
            patient.name = updates["name"].strip()
        if "dob" in updates:
            patient.dob = self._validate_dob(updates["dob"])
        if "gender" in updates:
            self._validate_gender(updates["gender"])
            patient.gender = updates["gender"]
        if "weight_kg" in updates:
            self._validate_weight(updates["weight_kg"])
            patient.weight_kg = float(updates["weight_kg"]) if updates["weight_kg"] is not None else None
        if "conditions" in updates:
            patient.conditions = self._validate_conditions(updates["conditions"]) or []
        if "extra" in updates:
            if not isinstance(updates["extra"], dict):
                raise ValueError("extra must be a dict")
            patient.extra.update(updates["extra"])

    def get_patient(self, patient_id: str) -> Dict[str, Any]:
        """Return a copy of patient data as a dictionary."""
        self._validate_patient_id(patient_id)
        if patient_id not in self._records:
            raise KeyError(f"Patient with id '{patient_id}' not found.")
        return copy.deepcopy(asdict(self._records[patient_id]))

    def remove_patient(self, patient_id: str) -> None:
        """Remove a patient record."""
        self._validate_patient_id(patient_id)
        if patient_id not in self._records:
            raise KeyError(f"Patient with id '{patient_id}' not found.")
        del self._records[patient_id]

    def list_patients(self) -> List[Dict[str, Any]]:
        """Return list of patient dicts."""
        return [copy.deepcopy(asdict(p)) for p in self._records.values()]

    # -------------------------
    # Statistical analysis
    # -------------------------
    def summary_stats(self, on_date: Optional[date] = None) -> Dict[str, Any]:
        """Return basic stats: count, avg/median age, gender counts, condition counts."""
        if on_date is None:
            on_date = date.today()
        patients = list(self._records.values())
        count = len(patients)
        ages = [p.age(on_date) for p in patients] if patients else []
        gender_counts = Counter(p.gender for p in patients)
        condition_counter = Counter(cond for p in patients for cond in p.conditions)

        stats = {
            "count": count,
            "average_age": round(mean(ages), 2) if ages else None,
            "median_age": median(ages) if ages else None,
            "age_min": min(ages) if ages else None,
            "age_max": max(ages) if ages else None,
            "gender_counts": dict(gender_counts),
            "condition_counts": dict(condition_counter),
        }
        return stats

    def age_distribution(self, bins: Optional[List[int]] = None, on_date: Optional[date] = None) -> Dict[str, int]:
        """
        Return counts of patients in age bins.
        bins: list of bin edges (e.g., [0, 18, 35, 50, 65, 120])
        returns mapping like '0-17': 5, '18-34': 10, ...
        """
        if on_date is None:
            on_date = date.today()
        if bins is None:
            bins = [0, 18, 35, 50, 65, 200]
        if sorted(bins) != bins or len(bins) < 2:
            raise ValueError("bins must be a sorted list with at least two edges")

        ages = [p.age(on_date) for p in self._records.values()]
        distribution: Dict[str, int] = {}
        for i in range(len(bins) - 1):
            lo, hi = bins[i], bins[i + 1]
            key = f"{lo}-{hi - 1}" if hi - 1 > lo else f"{lo}-{hi}"
            distribution[key] = sum(1 for a in ages if lo <= a < hi)
        return distribution

    # -------------------------
    # Utility / IO
    # -------------------------
    def to_dict(self) -> Dict[str, Dict[str, Any]]:
        """Return all records as a mapping patient_id -> dict."""
        return {pid: copy.deepcopy(asdict(p)) for pid, p in self._records.items()}

    def from_list(self, patients: List[Dict[str, Any]]):
        """Bulk add patients from a list of dicts (will validate)."""
        for p in patients:
            self.add_patient(
                patient_id=p["patient_id"],
                name=p["name"],
                dob=p["dob"],
                gender=p.get("gender"),
                weight_kg=p.get("weight_kg"),
                conditions=p.get("conditions"),
                extra=p.get("extra"),
            )
````

````python
# Instantiate the patient manager and demonstrate basic usage

mgr = PatientRecordManager()

# Add some sample patients (ISO date strings are accepted)
mgr.add_patient("p001", "Alice Smith", "1995-06-15", gender="F", weight_kg=65.0, conditions=["hypertension"])
mgr.add_patient("p002", "Bob Jones", "1980-01-20", gender="M", weight_kg=82.5, conditions=["diabetes"])
mgr.add_patient("p003", "Charlie Kim", "2010-09-05", gender="M", weight_kg=45.0, conditions=[])

# Show all patients
print("All patients:")
for p in mgr.list_patients():
    print(p)

# Summary statistics
print("\nSummary stats:")
print(mgr.summary_stats())

# Age distribution with default bins
print("\nAge distribution:")
print(mgr.age_distribution())

# Retrieve a single patient
print("\nPatient p002:")
print(mgr.get_patient("p002"))

# Update a patient and show the change
mgr.update_patient("p003", name="Charlie K.", weight_kg=47.0, conditions=["asthma"])
print("\nUpdated p003:")
print(mgr.get_patient("p003"))

# Remove a patient
mgr.remove_patient("p001")
print("\nAfter removing p001, remaining patient IDs:", [p["patient_id"] for p in mgr.list_patients()])
````
